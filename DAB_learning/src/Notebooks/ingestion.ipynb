```python
dbutils.widgets.text("catalog_name", "default_value", "My Widget Label")
dbutils.widgets.text("schema_name", "default_value", "My Widget Label")

# To retrieve the widget value later:
# value = dbutils.widgets.get("my_widget")
```

```python
catalog=dbutils.widgets.get("catalog_name")
schema=dbutils.widgets.get("schema_name")
```

```python

df=spark.sql(f'select * from {catalog}.{schema}.columns')
```

```python
display(df)
```

```python
new_df=df.select('table_catalog','table_schema','table_name','column_name','comment')
new_df.write.mode('overwrite').option('mergeschema','true').saveAsTable(f'{catalog}.{schema}.data')
```
